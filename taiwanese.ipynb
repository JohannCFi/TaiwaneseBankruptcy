Ces deux colonnes devraient être l'inverse l'une de l'autre or on voit que leur produit ne vaut pas 1 donc cela indique qu'elles ont peut être été transformées. Dans tous les cas, elles donnent la même information donc une des deux devra forcément être supprimée. On vérifie ça avec la corrélation.

```python
print(np.corrcoef(data["Equity to Liability"],data["Liability to Equity"])) 
# corrcoef ne calcule que lacorrélation de Pearson
```

Pas proche de -1 car sans paramètre on utilise la corrélation de Pearson qui mesure une relation linéaire. Or une relation inverse n'est pas linéaire, on calcule la corrélation de spearman qui s'intéresse seulement au fait les valeurs évoluent ensemble ou non.


```python
# Corrélation de Spearman 
data[["Equity to Liability","Liability to Equity"]].corr(method="spearman")
```

Effectivement très proche de -1 donc on regarde s'il n'y a pas d'autres features dans ce cas.

```python
Spearman_corr = data.drop(columns="Net Income Flag").corr(method="spearman")
Spearman_corr
```

Ma requête qui me semblait redondante puisque qu'on ne supprime ni la diagonale (qui vaut toujours 1) ni les doublons :


```python
paires = Spearman_corr.stack()
paires[paires.abs() > 0.9].sort_values()
```

Celles de Claude qui fait ce que je ne savais pas faire grâce à triu qui met False partout (dans la matrice remplie de True) en dessous de la diagonale (sans la diagonale car k=1). On supprime donc les doublons et la diagonale (le reste de la requête fonctionne comme la précédente):

```python
mask = np.triu(np.ones(Spearman_corr.shape, dtype=bool), k=1)   # triangle supérieur strict
paires = Spearman_corr.where(mask).stack()
paires[paires.abs() > 0.9].sort_values(key=abs, ascending=False)
```
